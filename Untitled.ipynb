```python
import pandas as pd
```

```python
df = pd.read_csv("output/training_data.csv")
```

```python
# find rows with empty data
empty_rows = df[df.isnull().any(axis=1)]
print(empty_rows)
```

```python
empty_rows = df.where(df == 0)
print(empty_rows)
```

```python
empty_or_zero_rows = df[(df.isnull().any(axis=1)) | (df.eq(0).any(axis=1))]
print(empty_or_zero_rows)
```

```python
df2 = df[['SALE DATE', 'BLOCK', 'LOT', 'ZIP CODE', 'SALE PRICE']]
```

```python
empty_or_zero_rows = df2[(df2.isnull().any(axis=1)) | (df2.eq(0).any(axis=1))]
print(empty_or_zero_rows)
```
